```python
import os


## Gives all files a prefix, suffix and new extension
def mass_rename_files(folder_path, prefix=None, suffix=None, new_extension=None):
    """Rename files in a directory with an optional prefix, suffix, or new extension."""
    # List all files in the directory
    for filename in os.listdir(folder_path):
        # Skip directories
        if os.path.isdir(os.path.join(folder_path, filename)):
            continue

        # Get the file's base name and extension
        base_name, ext = os.path.splitext(filename)

        # Build the new filename with optional prefix, suffix, and extension
        new_name = base_name
        if prefix:
            new_name = f"{prefix}{new_name}"
        if suffix:
            new_name = f"{new_name}{suffix}"
        if new_extension:
            ext = new_extension  # Replace extension

        # Construct the full paths
        old_file = os.path.join(folder_path, filename)
        new_file = os.path.join(folder_path, f"{new_name}{ext}")

        # Rename the file
        os.rename(old_file, new_file)
     #   print(f"Renamed: {old_file} --> {new_file}")

if __name__ == "__main__":
    # Specify the folder where the files are located
    FOLDER_PATH = "/Users/kyriakos.trakakis/Downloads/FUGIN_data"

    # Rename files by adding a prefix, suffix, or changing the extension
    PREFIX = ""  # Example: add "prefix_" at the beginning of each filename
    SUFFIX = ""  # Example: add "_suffix" at the end of each filename
    NEW_EXTENSION = ".fits"  # Example: change the file extension to ".txt"
    
  #  mass_rename_files(FOLDER_PATH, prefix=PREFIX, suffix=SUFFIX, new_extension=NEW_EXTENSION)
```

```python
def rename_fits_files(folder_path):
    # Modified for rms files, careful!
    """Rename FITS files to the format 'FGN_046_13CO.fits'."""
    for filename in os.listdir(folder_path):
        if filename.endswith(".fits"):
            base_name, ext = os.path.splitext(filename)

            # Split the base name by underscores
            parts = base_name.split("_")
            
            # Check if the filename starts with 'FGN' and has enough parts
            if len(parts) >= 6 and parts[0] == "FGN":
                # Extract the 'FGN_046', '13CO' part (3-digit identifier and molecule)
                new_base_name = f"{parts[0]}_{parts[1][:3]}_{parts[3]}_{parts[6]}"

                # Rename the file
                old_file = os.path.join(folder_path, filename)
                new_file = os.path.join(folder_path, new_base_name + ext)

                os.rename(old_file, new_file)
                print(f"Renamed: {old_file} --> {new_file}")

if __name__ == "__main__":
    # Specify the folder where the FITS files are located
    FOLDER_PATH = "/Users/kyriakos.trakakis/Downloads/Mopra_FGN_project/FGN"

    # Run the renaming function
   # rename_fits_files(FOLDER_PATH)
```

```python
import subprocess

def run_command(command, cwd=None):
    """Run a command and print its output."""
    print(f"Running command: {command}")
    result = subprocess.run(command, shell=True, cwd=cwd, text=True, capture_output=True)
    if result.returncode != 0:
        print(f"Error running command: {result.stderr}")
    else:
        print(result.stdout)

FOLDER_PATH = "/Users/kyriakos.trakakis/Downloads/Mopra_FGN_project/FGN"
for filename in os.listdir(FOLDER_PATH):
        if filename.startswith("FGN"):
            # Get the filename
            fits_file = os.path.join(FOLDER_PATH, filename)
            base_name = os.path.splitext(filename)[0]

            # Turn to ndf
            ndf_file = os.path.join(FOLDER_PATH, f"{base_name}.sdf")
            run_command(f"  /usr/local/star-2023A/bin/convert/fits2ndf {fits_file} {ndf_file}")
```
