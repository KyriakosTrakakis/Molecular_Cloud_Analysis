```python
# Initialisess stuff
# Run fearlessly

os.chdir(clusters_path)

# Constants
a1, a2, a3 = 1.0077, 0.0394, 0.0071  # Brand & Blitz (1993)
theta0     = 236.0              # km/s
R0         = 8.15               # kpc
omega0     = theta0 / R0
dtor       = pi / 180

# Initialize outputs
n              = len(df)
R              = np.zeros(n)
Vcalc          = np.zeros(n)
d_near         = np.zeros(n)
d_far          = np.zeros(n)
distance       = np.full(n, np.nan)
dist_method    = np.zeros(n, dtype=int)
```

```python
# Full rotation curve code
# Run fearlessly

# Step 1: Rotation Curve + Near/Far Distance Calculation
for i in range(n):
    l, b, V = df['x_cen'][i], df['y_cen'][i], df['v_cen'][i]/1000
    
    Ri = 0.1
    count = 0
    while True:
        count +=1
        Ri += 0.01
        wi = omega0 * ((a1 * (Ri/R0)**(a2 - 1)) + a3 * (R0 / Ri))
        Vtest = R0 * (wi - omega0) * sin(radians(l)) * cos(radians(b))
        if abs(Vtest - V) < 0.4:
            break

        if count > 1e6:
            count = 0
            Ri = np.nan
            break
    
    R[i] = Ri
    Vcalc[i] = Vtest

    A = cos(radians(b))**2
    B = 2 * R0 * cos(radians(b)) * cos(radians(l))
    C = R0**2 - Ri**2
    delta = B**2 - 4 * A * C

    if delta <= 0:
        d1 = d2 = R0 * cos(radians(l)) / cos(radians(b))
    else:
        sqrt_delta = sqrt(delta)
        d1 = -( -B + sqrt_delta ) / (2 * A)
        d2 = -( -B - sqrt_delta ) / (2 * A)

    d_near[i], d_far[i] = min(d1, d2), max(d1, d2)

    # Outer Galaxy check
    if R[i] >= R0 or d_near[i] <= 0:
        distance[i] = d_far[i]
        dist_method[i] = 2  # Outer Galaxy
    #print(i)

# Step 2: Tangent Point Method
# Apply only to unresolved sources (dist_method = 0)
for i in range(n):
    if dist_method[i] != 0:
        continue
    l = df['x_cen'][i]
    V = df['v_cen'][i]
    tan_vel = theta0 * sin(radians(l))
    if abs(V - tan_vel) < 10 and (d_far[i] - d_near[i]) <= 1.0:
        distance[i] = (d_near[i] + d_far[i]) / 2
        dist_method[i] = 3  # Tangent method

# Step 3: Scale Height Method
# Still unresolved sources
for i in range(n):
    if dist_method[i] != 0:
        continue
    b = df['y_cen'][i]
    z = d_far[i] * 1000 * tan(radians(b))  # pc
    if abs(z) > 120:
        distance[i] = d_near[i]
        dist_method[i] = 4  # Scale height method

# Step 4: Append Results
df['R_GC'] = R
df['d_near'] = d_near
df['d_far'] = d_far
df['distance'] = distance
df['dist_method'] = dist_method
```

```python
# Function definition
# Run fearlessly

def determine_distance_assignment(txt_path, d_near, d_far):
    components = []

    with open(txt_path, 'r') as f:
        lines = f.readlines()

    # Filter lines that contain "Probability component"
    prob_lines = [line.strip() for line in lines if line.strip().startswith("Probability component")]

    for line in prob_lines:
        parts = line.split()
        try:
            dist = float(parts[3])
            prob = float(parts[5])
            components.append((dist, prob))
        except (IndexError, ValueError):
            continue

    if not components:
        raise ValueError("No valid probability components found.")

    # Get distance of component with highest probability
    best_dist, _ = max(components, key=lambda x: x[1])
    
    all_distances = [comp[0] for comp in components]
    min_distance = min(all_distances)
    max_distance = max(all_distances)

    if abs(best_dist - d_near) <= abs(best_dist - d_far):
            return 'near'
    else:
            return 'far'

prt_dir = '/Users/kyriakos.trakakis/Downloads/v2.4.1_bundle/'  # Update if needed
final_distances = []
distance_flags = []  # Optional: stores 'near' or 'far'
```

```python
# Selects near or far distances
# Run fearlessly

for idx, row in df.iterrows():
    obj_id = int(row['_idx'])
    prt_file = os.path.join(prt_dir, f"{obj_id}.prt")

    if row['R_GC'] < R0:

        try:
            choice = determine_distance_assignment(prt_file, row['d_near'], row['d_far'])
            if choice == 'near':
                final_distances.append(row['d_near'])
                distance_flags.append('near')
            else:
                final_distances.append(row['d_far'])
                distance_flags.append('far')
        except Exception as e:
            print(f"Warning: Failed to process {prt_file} due to {e}")


            if row['d_near'] < 0:
                if row['d_far'] > 0:
                    final_distances.append(row['d_far'])
                    distance_flags.append('far')
                elif row['d_far'] < 0:
                    final_distances.append(min(abs(row['d_near']), abs(row['d_far'])))
                    distance_flags.append('near')
                

    else:
        final_distances.append(row['d_far'])
        distance_flags.append('No_KDA')

df['distance'] = final_distances
df['distance_flag'] = distance_flags  # Optional for debugging
df
```

```python
## Creates pv plot by averaging over a positional coordinate. Saves a plot with clever naming so it doesn't overwrite 
## wrong stuff. Specify "type" bv or lv. Copied from BDC.
df_2 = pd.read_csv("/Users/kyriakos.trakakis/Downloads/FUGIN_data/13CO/All_FGN_cloud_catalogue_complete.csv")

l_bounds = [40,50]
region = f"{l_bounds[0]}-{l_bounds[1]}"
filename = f'FGN_13CO_{region}_23arcsec_trimmed.fits'
file_path = "/Users/kyriakos.trakakis/Downloads/FUGIN_data/13CO/"
fits_file = file_path + filename
destination_path = file_path
type = "lv"
l = df_2['x_cen']
v = df_2['v_cen']
match = df_2['has_match']


mask = (l < l_bounds[1]) & (l > l_bounds[0])

l = l[mask]
v = v[mask]
match = match[mask]



def make_pv_plot_with_points(
    fits_file,
    save_fits,
    pv_type="lv",
    method="average",
    show=True,
    vmin=None,
    vmax=None
):
    """
    Create LV or BV PV diagrams from a (v, b, l) FITS cube.

    FITS axes:
        1 = Galactic longitude
        2 = Galactic latitude
        3 = velocity

    NumPy array:
        data[v, b, l]

    method:
        "average"   -> mean intensity (K)
        "integrate" -> spatially integrated intensity (K deg)
    """

    if pv_type not in ("lv", "bv"):
        raise ValueError("pv_type must be 'lv' or 'bv'")

    if method not in ("average", "integrate"):
        raise ValueError("method must be 'average' or 'integrate'")

    # Read FITS cube
    with fits.open(fits_file) as hdul:
        data = hdul[0].data.astype(float)
        header = hdul[0].header.copy()

    if data.ndim != 3:
        raise ValueError("Input FITS file must be 3D")

    nv, nb, nl = data.shape
    print(f"Cube shape (v,b,l): {nv}, {nb}, {nl}")

    # Select spatial axis to collapse
    if pv_type == "lv":
        collapse_axis = 1
        x_axis = 1
        xlabel = "Galactic Longitude (deg)"
        spatial_fits_axis = 2
    else:
        collapse_axis = 2
        x_axis = 2
        xlabel = "Galactic Latitude (deg)"
        spatial_fits_axis = 1

    # Collapse cube
    with np.errstate(invalid="ignore"):
        if method == "average":
            pv = np.nanmean(data, axis=collapse_axis)
            cbar_label = "Mean Brightness Temperature (K)"
        else:
            # Spatial integration in degrees
            spatial_step = abs(
                (header[f"CDELT{spatial_fits_axis}"]
                 * u.Unit(header.get(
                     f"CUNIT{spatial_fits_axis}", "deg"
                 ))).to_value(u.deg)
            )

            pv = np.nansum(data, axis=collapse_axis) * spatial_step
            cbar_label = "Spatially Integrated Intensity (K deg)"

    print("PV shape (velocity, position):", pv.shape)

    # Display limits
    if vmin is None:
        vmin = 0.0

    if vmax is None:
        vmax = np.nanpercentile(pv, 99)

    # Construct PV FITS header
    new_header = fits.Header()

    new_header["NAXIS"] = 2
    new_header["NAXIS1"] = pv.shape[1]
    new_header["NAXIS2"] = pv.shape[0]

    # Spatial axis
    # Linear angular coordinate rather than celestial projection
    spatial_unit = u.Unit(header.get(f"CUNIT{x_axis}", "deg"))

    new_header["CTYPE1"] = "LINEAR"
    new_header["CRVAL1"] = (
        header[f"CRVAL{x_axis}"] * spatial_unit
    ).to_value(u.deg)
    new_header["CRPIX1"] = header[f"CRPIX{x_axis}"]
    new_header["CDELT1"] = (
        header[f"CDELT{x_axis}"] * spatial_unit
    ).to_value(u.deg)
    new_header["CUNIT1"] = "deg"

    # Velocity axis
    velocity_unit = u.Unit(header.get("CUNIT3", "m/s"))

    new_header["CTYPE2"] = header.get("CTYPE3", "VELO-LSR")
    new_header["CRVAL2"] = (
        header["CRVAL3"] * velocity_unit
    ).to_value(u.km / u.s)
    new_header["CRPIX2"] = header["CRPIX3"]
    new_header["CDELT2"] = (
        header["CDELT3"] * velocity_unit
    ).to_value(u.km / u.s)
    new_header["CUNIT2"] = "km/s"

    # Preserve spectral metadata
    for key in ("SPECSYS", "RESTFRQ", "RESTFREQ"):
        if key in header:
            new_header[key] = header[key]

    new_header["BUNIT"] = (
        "K" if method == "average" else "K deg"
    )

    new_header["BTYPE"] = f"{pv_type.upper()} PV map ({method})"

    # ------------------
    # Save FITS
    # ------------------
    if save_fits:
        fits.writeto(pv_name, pv, new_header, overwrite=True)
        print(f"Saved PV FITS to: {save_fits}")

    # ------------------
    # Plot
    # ------------------
    if show:
        pv_wcs = WCS(new_header)

        fig, ax = plt.subplots(
            figsize=(8, 6),
            dpi=200,
            subplot_kw={"projection": pv_wcs}
        )

        im = ax.imshow(
            pv,
            origin="lower",
            aspect="auto",
            cmap="gray_r",
            vmin=vmin,
            vmax=vmax
        )

        ax.set_xlabel(xlabel)
        ax.set_ylabel("Velocity (km s$^{-1}$)")
        ax.invert_yaxis()

        print("Total points:", len(l))
        print("Matched points:", np.sum(match))
        print("Unmatched points:", np.sum(~match))

        print("Matched coordinates:", np.shape(l[match]), np.shape(v[match]))
        print("Unmatched coordinates:", np.shape(l[~match]), np.shape(v[~match]))

        if np.sum(~match >0):
            ax.scatter(
                l[~match],
                v[~match],
                s=4,
                color='blue',
                alpha=0.5,
                label = 'No Match',
                transform=ax.get_transform('world')
)
        if np.sum(match) > 0:
            ax.scatter(
                l[match],
                v[match],
                s=10,
                color='red',
                edgecolor='black',
                label = 'ATLASGAL match',
                transform=ax.get_transform('world')
)
        ax.legend()
        
        divider = make_axes_locatable(ax)
        plt.colorbar(im, ax=ax, label=cbar_label)

        plt.tight_layout()
        plt.savefig(f"{type}_plot")
        plt.show()

    return pv, new_header


# ------------------
# Example call
# ------------------
pv_name = fits_file.replace(".fits", "_"+type+".fits")
pv, hdr = make_pv_plot_with_points(
    fits_file,
    save_fits= pv_name,
    pv_type=type,
    method="average",
  #  save_fits = None,

)
```

```python
plt.figure(figsize=(8,5))

def equal_count_mass_function(values):

    values = values[values > 0].dropna().values
    N = len(values)
    num_bins = math.ceil(2 * (N ** 0.4))

    sorted_vals = np.sort(values)

    # Quantile bin edges
    quantiles = np.linspace(0, 1, num_bins + 1)
    bin_edges = np.quantile(sorted_vals, quantiles)
    bin_edges = np.unique(bin_edges)

    num_bins_actual = len(bin_edges) - 1

    counts = np.histogram(sorted_vals, bins=bin_edges)[0]
    bin_widths = bin_edges[1:] - bin_edges[:-1]
    density = counts / bin_widths

    # Compute bin medians
    bin_values = []
    for i in range(num_bins_actual):
        left = bin_edges[i]
        right = bin_edges[i+1]

        mask = (sorted_vals >= left) & (sorted_vals < right)
        if i == num_bins_actual - 1:
            mask = (sorted_vals >= left) & (sorted_vals <= right)

        bin_values.append(sorted_vals[mask])

    bin_medians = np.array([np.median(vals) for vals in bin_values])

    return bin_medians, density

col_labels = [
    'M_pixel_Sol_13CO',
    'M_pixel_Sol_12CO',
    'M_LTE_Sol'
]

colors = [
    '#000000',  # black
    '#E69F00',  # orange
    '#56B4E9'   # sky blue
]

styles = [
    '-',                  # solid
    '--',                 # dashed
    '-.'                  # dash-dot
]

start_indices = [
    17,
    17,
    17
]

for i in range(len(col_labels)):

    col_label = col_labels[i]

    bin_medians, density = equal_count_mass_function(df_2[col_label])

    bin_medians_log = np.log10(bin_medians)

    # ---- Select high-mass tail ----
    start_index = start_indices[i]   # change if needed
    x_log = bin_medians_log[start_index:]
    y = density[start_index:]
    y_log = np.log10(y)

    # ---- Fit in log-log space ----
    coeffs, cov = np.polyfit(x_log, y_log, 1, cov=True)
    slope = coeffs[0]
    intercept = coeffs[1]

    m_err = np.sqrt(cov[0, 0])
    b_err = np.sqrt(cov[1, 1])


    a = slope
    A = 10**intercept

    # ---- Predicted values ----
    fit_y = A * (10**bin_medians_log)**(a)

    # ---- χ² (linear space) ----
    predicted_tail = A * (10**x_log)**(a)
    chi2 = np.sum((y - predicted_tail)**2 / predicted_tail)

    print(f"\nColumn: {col_label}")
    print(f"A = {A:.4e}")
    print(f"a = {a:.4f} ± {m_err:.4f}")
    print(f"χ² = {chi2:.4f}")

    # ---- Plot ----
    plt.scatter(bin_medians, density, alpha=0.7, s=3, color = colors[i])
    plt.plot(bin_medians, fit_y,
             linestyle=styles[i], color=colors[i],
             label=f"{col_label}  (a={-a:.2f})")


plt.xscale("log")
plt.yscale("log")
plt.xlabel(r"Cloud Mass $(M_\odot)$")
plt.ylim([0.00001, 20])
plt.xlim([80, 400000])
plt.ylabel(r"$\Delta N / \Delta M$")
plt.plot([], [], ' ', label='Fit: $A\ M^{-a}$')
plt.legend()
plt.grid(True, which="both", ls="", alpha=0.6)
#plt.savefig(destination_path + "6_power_laws.png")
```
